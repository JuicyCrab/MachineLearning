### Dataset API 

- Sklearn offers its own datasets through the dataset API 
    - import sklearn.datasets 

- The three interfaces to get dataset depending on user's needs 
    - Loaders 
        - used more for the smaller standard datasets 
    - Fetchers 
        - loads and downloads datasets 
    Loaders and fetchers return a bunch object
        - an array of the n_features * n_samples with data 
        - numpy array length of samples, containing the target and key values 
    Dataset generating functions 
        - generate controlled synthetic datasets 
        - functions return a tuple (X, y)
            + consist of n_samples * n_features 
            + Numpy array X and array length of n_samples containing the targets y 
 

### Cross Validation 

Using cross validation to evaluate the binary classifier model 
    - For this instance there are 3 folds 
        + Coding implementation 

            from sklearn.model_selection import cross_val_score 
            cross_val_score(classifier object, X_train, y_train, cv=3, scoring="accuracy")

            + Parameters 
                1. cv is the amount of folds for cross validation 
                2. scoring is the type of report wanting to be returned from the function 
    Possible question/s of cross validation 
    1. How is the amount of folds known? 
        - Unfortunately, there is no direct equation to find the optimal number, but it typically ranges from 5-10 folds. 
          The reasoning is because the test set for the validation set should have enough variation depending on the dataset and use at least 20 % of your training data. Also, you should not put the fold number as the amount of instances, since it will be computationally expensive. If your dataset is on the smaller end, it might be better to consider Leave-one-out-cross validation.
