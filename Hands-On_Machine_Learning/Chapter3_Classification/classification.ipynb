### Cross Validation 

- Cross Validation to evaluate the binary classifier model 
    - For this instance there are 3 folds 
        + Coding implementation 

            > from sklearn.model_selection import cross_val_score <br>
            > cross_val_score(classifier object, X_train, y_train, cv=3, scoring="accuracy")

            + Parameters 
                1. cv is the amount of folds for cross validation 
                2. scoring is the type of report wanting to be returned from the function 
- Possible question/s of cross validation 
    1. How is the amount of folds known? 
        - Unfortunately, there is no direct equation to find the optimal number, but it typically ranges from 5-10 folds. 
          The reasoning is because the test set for the validation set should have enough variation depending on the dataset and use around 20 % of your training data. Also, you should not put the fold number as the amount of instances, since it will be computationally expensive and low bias. If your dataset is on the smaller end and estimations are critical, it might be better to consider Leave-one-out-cross validation. 

        - Leave One Out Cross Validation 
            + Extreme case were the fold size is the training size

### Confusion Matrices 
- Confusion matrices are a way to visualize the the number of times one class is classified is to another 
- They consist of four types of predictions: 
    - For the examples of each type the scenario of predicting if a patient has cancer or not. The positive class will be having cancer and the negative is not having cancer.
    <ol>
        <li>True Positives</li>
        + These are the classifications that correctly predicted a positive case <br>
        + Classifier diagnosing a user to have cancer, which they actually do have cancer 
        <li>True Negatives</li>
        + These are the predictions were the prediction is positive but is really not. <br>
        + Classifier predicting someone has cancer but they don't actually do.
        <li>False Positives</li>
        + Predictions were the negative class is correctly predicted <br>
        + Predicting someone does not have cancer and actually doesn't 
        <li> False negatives </li>
        + A prediction were the negative false is predicted but is actually positive <br> 
        + Predicting someone doesn't have cancer but they do
    </ol>
- How to use:
> from sklearn.metrics import confusion_matrix <br>
> cm = confusion_matrix(y_train_class, y_train_pred) 
- Returns an array having each row as the actual class and the column as the predicted class 

    

### Concise Performance Measures than Confusion Matrices 

- Legend:
  + TP = True Positives 
  + TN = True Negatives
  + FP = False Positives 
  + FN = False Negatives 

- Accuracy 
  + The percentage of true predictions from the total

  + $\frac{TP + TN}{TP + TN + FP + FN}$<br>
- Precision 
  + The accuracy of the positive predictions 
  + $\frac{TP}{ TP + FP}$
  + A trivial thing with precision is the case where there might only be one positive case from creating a model only to predict negative cases. The result could be 100%, but that's only because of the single instance, which can misrepresent how the model could really be classifying. 

- Recall 
  + This is typically used with precision to avoid the trivial case. 
  + This also known as the sensitivity of precision or the true positive rate, since it displays the true ratio of positive instance correctly detected 
  + $\frac{TP}{TP + FN}$

- Precision and Recall Code 
> from sklearn.metrics import precision_score, recall_score <br>
> precision_score(y_train_actual, y_train_pred) <br>
> recall_score(y_train_actual, y_train_pred) <br>


- F1 Score 
 + Combines precision and recall into a single score 
 + Best score at 1 and worst at 0 
 + Equation derivation: <br>
  F1 = $\frac{2}{\frac{1}{precision} + \frac{1}{recall}}$ = 2 * $\frac{precision * recall}{precision + recall}$ = $\frac{TP}{TP + \frac{FN + FP}{2}}$
> from sklearn.metrics import f1_score <br>
> f1_score(y_true, y_pred)

  - Trade offs to consider 
    + Unfortunately, precision and recall can't both be high at the same time but a single one can with the expense of the other being lower 
    + An example is when creating thresholds that decides if a a classification is positive or negative 
    + If the positive class has a lower threshold, this increases recall but decreases precision  

### Extra Tools for Classifiers 

- Precision Recall Curve 
    + A visualization of precision and recall in a graph setting 
    > from sklearn.metrics import precision_recall_curve <br>
    > precisions, recalls, thresholds = precision, recall, thresholds = precision 
- Receiver Operating Characteristic 
    + Plots the recall against the false positive rate or also known as the fall-out 
        + fall out is the ratio of negative instances that are incorrectly classified as positive and is equal to 1 - true negative rate 
        + true negative rate is called specificity 
        + y axis is the sensitivity(TPR) and x axis is the specificity(TNR)
        + This graph showcases combinations of thresholds and the corresponding sensitivity and specificity 
    > from sklearn.metrics import roc_curve  <br>
    > fpr, tpr, thresholds = roc_curve(y_train_actual, y_pred)
    + trade offs
        1. The higher the recall or TPR the more false positives the classifier predicts 
    + Also, there is a linear line in te graph which represents a totally random classifier so its best for the line being in the tope left corner 


-  When to use Precision Recall Curve or ROC 
    + The PRC whenever the positive class is rare or the false positives are more important the false negatives


### Multiclass Classification/ Multinomial Classification 
- Multiclass and multinomial classification 
    + A type of classification that classifies more than 2 things 

- One versus rest (OVR) / one versus all 
    + A classification method where there is a binary classifier for every option 
    + For each training instance there are some number binary classifiers that have the decision of being classified something or not that 
    + Ex: classifying numbers 0-10, there are 11 classifiers and each classifier and the one versus all will basically say is this a one or not 
    + That question will be applied to those other numbers as well 

- One versus one (ONO)
    + A strategy where the classification will classify as one option or another option 
    + Ex: using the number example above its the same as saying is the input data is a 1 or 2 and this will continue for the other combinations of this number or that number 
    + The advantage is that each classifier only needs to be trained on the part of the training set containing the two classes that it must distinguish 

- Sklearn 
    + Luckily, sklearn automatically figures out when to use OVR or ONO
