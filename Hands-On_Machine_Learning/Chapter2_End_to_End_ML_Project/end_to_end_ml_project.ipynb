#### 3. Selecting Performance Measures 
- Root Mean Square Error 
    - RMSE(**X**, h) = $`\sqrt{\frac{1}{m} \sum_{i=1}^{m}(h(\textbf{x}^{i})-y^{(i)})^{2}}`$
        - m = amount of training instances or observations 
        - $`h(\textbf{x}^{i})`$ = value outputted from the model given the features of the training instance
        - $`y^{(i)}`$ = actual value from the label 
        - Preferred for regression tasks 

- Mean Absolute Error 
    - MAE (**X**, h) = $`\frac{1}{m} \sum_{i=1}^{m}\left|(h(\textbf{x}^{i})-y^{(i)})\right|`$
        - Instead of squaring the difference from the actual and the model's result, the absolute difference is just used
        - This is typically used when there are more outliers because RMSE is more sensitive, given that it uses squared values.
- #### **Vocab**
    - RMSE and MAE measure distance between 2 vectors. 
        1. RMSE uses Euclidean distance or l2 norm 
            - The vector distance from the origin to the head of the final vector 
        2. MAE uses Manhatten distance/ l1 norm
            - This is the summation of the distance of all directions 
                - The Manhatten name is relating to the walking of going orthogonal directions in the city of Manhatten

### 6. Prepare and Clean the Data 

- For supervised learning tasks separate the labels from the data 
- Cleaning the Data 
    1. Missing Values
        - Some methods are by using dropna(), drop(), and fillna() methods 
        or 
        SkLearn SimpleImputer Class 
        - from sklearn.impute import SimpleImputer 
        - imputer = SimpleImputer(strategy="median/mode/mean")

    2. Textual/Categorical values 
        - ML models are way better with numerical values, but data can be given in the form of text 
        - Encoding helps put textual values into numbers 
        - If the text has order 
            Ex: small, medium, large = 0, 1, 2
            - Use OrdinalEncoder 
            - Coding implementation
                - from sklearn.preprocessing import OrdinalEncoder 
                - ordinal_encoder = OrdinalEncoder()
                - encoded_vals = ordinal_encoder.fit_transform(categorical_data) -> returns an array 
        - IF the text doesn't have a specific order 
            - Use OneHotEncoder 
            - from sklearn.preprocessing import OneHotEncoder 
            - one_hot_encoder = OneHotEncoder()
                - possible parameters are sparse_output(Boolean)(returns the encoding in compressed form to remove excessive 0s)
            - encoded_column = one_hot_encoder.fit_transform(categorical_values) -> returns a binary(0,1) column 
        

### 7. Feature Scaling and Transformation 

- Machine Learning models perform way better when the numerical values are consistently scaled 
- Methods for scaling 
    1. Min-Max Scaler or also known as normalization
        - values are scaled are ranged from 0 and 1 
        - formula = $`\frac{value - min}{max - min}`$
        - Code 
            - from sklearn.preprocessing import MinMaxScaler 
            - min_max = MinMaxScaler()
            - num_min_max = min_max.fit_transform(data)
    2. Standard Scaler 
        - no particular range like Min-Max 
        - formula = $`\frac{value - \bar{x}}{\sigma}`$
        - This is the same as z-score 
        - mean = 0 & standard deviation = 1 
        - less affected with outliers 
        - Code 
            - from sklearn.preprocessing import StandardScaler 
            - scaler = StandardScaler()
            - scaled_data = scaler.fit_transform(data_wanting_to_be_scaled)
- Before applying the scaler it's best to get the data values symmetrical 
    - For example if there is a distribution towards more rare values like a power law distribution its better to normalize the distribution
