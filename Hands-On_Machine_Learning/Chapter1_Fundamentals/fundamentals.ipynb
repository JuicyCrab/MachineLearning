## Chapter 1: Fundamentals

### Types of Machine Learning Systems 
- Supervised 
    - Occurs when the labels or desired solutions are known for the data 
        - Typically used for classification tasks 
            Ex: When training a spam class filter during training each message has a label of spam or not spam 
    - Target is synonymous with label  
- Unsupervised 
    - Training data is unlabeled 
    - The system tries to learn without a teacher 
    - Clustering algorithms are examples
- Reinforcement 
    - Very different by using an agent that can observe the environment, select and perform actions, and get rewards, penalties, or both in turn 
    - The system must learn by creating a policy that maximizes rewards and minimizes penalties 
        - A policy defines the agent what action needs to be taken given the situation 
    -Ex: A robot learning how to walk 

### Offline(Batch) Learning

 - Type of learning that uses up all the available data.
 - A disadvantage for using all the data at once can prevent a system from learning incrementally.
 - The typical process 
    - System is trained and launched into production 
    - When launched the system can't learn anymore and just applies what is has learned. 


### Online Learning 

- Train a system incrementally with mini-batches 
    - mini-batches are samples from a large dataset 
- Normal implementations are for scenarios for adapting quick (Stock market models)
- Learning rate parameter
    - Tells the system how fast to adapt to changing data 
    - A higher learning rate means the system will learn the new data more, which leads to quickly forgetting the previous data.
    - A smaller learning rate tells the system to learn slower. This allows the system to not be as sensitive to new data and outliers.
- For online learning, if bad data is given to the system, the system's performance will gradually decline.

### Categorizing ML Systems

#### Instance Based 

- The machine learns by storing all instances of data and then using these instances to make predictions on new data. 
- The machine compares the new data to the instances it has seen before and uses the closest match to make a prediction.
    - measure of similarity and no model is created 
- Used in pattern recognition, clustering, and anomaly detection. 
- Disadvantages include slower results, less accuracy since it can have a smaller dataset, and limited understanding of data. 


#### Model Based Learning 

- A system uses the dataset to create a model that can predict outcomes based on input data.
- The model is typically created using statistical algorithms such as linear regression, logistic regression, decision trees, and neural networks.
- A disadvantage is the requirement of a large dataset.
- article: [Instance vs Model Based Learning](https://medium.com/@pp1222001/model-based-vs-instance-based-learning-understanding-the-differences-with-examples-1545c9c3a056)

### Challenges of ML Projects 

####  Data 

##### 1. Insufficient Quality of Data 
- A toddler has the ability to learn what an apple is by someone pointing to one and saying apple.
- Machine learning is not at the stage yet and needs large sets of data to rely on.

##### 2. Non-Representative Training Data
- In order to generalize well it is important the training data reflects representative of the new cases you want to generalize to.
    - This applies to instance and model learning 
- If the sample is to small there is sampling noise 
    sampling noise: nonrepresentative data as result of chance 
- Large datasets can be nonrepresentative if the sampling method is flawed, which is called sampling bias
    - When using data always ask yourself is there an underlying pattern that should be known that can interrupt generalization 

##### 3. Poor Quality Data
- Uncleaned data can influence the model's recognized patterns 
- Examples of poor quality could be outliers and missing values
    - For missing values the best approach to fix is replacing the values with the mean, median, mode, or specific value

##### 4. Irrelevant Features 
- Your system will only be capable of learning if the training data contains enough relevant features and not too many irrelevant ones.
- Garbage in, Garbage out 
Solutions 
- Feature selection
    - Selecting the needed features 
- Feature extraction 
    - Combining existing features to produce a more useful one
        - Dimensionality reduction algorithms can help 
- Creating new features by gathering new data  

##### 5. Overfitting the Training Data 

- Overfitting: Is the idea of a model performing well on the training data but not on the unseen data also is **Overgeneralizing**
    - Must know 
- Ways to fix 
    - Simplify the model to fewer parameters, reducing number of attributes, or by constraining the model 
    - Gather more training data 
    - Reduce the noise in the training data(remove outliers and fix data errors)


##### 5. Underfitting the Training Data 
- Underfitting: Occurs when a model is too simple and can't learn patterns from the data 
- Fixes 
    - Select a more powerful model, with more parameters 
    - Feed better features to the learning algorithm 
    - Reduce the constraints on the model(reducing regularization hyperparameter)


#### No Free lunch theorem 

- A common phrase pointing out that no model is best for everything.
